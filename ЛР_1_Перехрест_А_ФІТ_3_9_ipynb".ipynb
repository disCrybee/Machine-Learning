### Завдання 25
Дано два вектори:

```text
a = [1, 2]
b = [2, 1]
```

Обчисліть косинус кута між ними за формулою через скалярний добуток і норми векторів.

Додатково знайдіть сам кут у градусах.

````python
# Код студента
import numpy as np

a = np.array([1, 2])
b = np.array([2, 1])

dot_product = np.dot(a, b)

norm_a = np.linalg.norm(a)
norm_b = np.linalg.norm(b)

cos_angle = dot_product / (norm_a * norm_b)

angle = np.degrees(np.arccos(cos_angle))

print("Косинус кута:", cos_angle)
print("Кут у градусах:", angle)
````

### Завдання 28
Для матриці:

```text

     [2, 6]]
```

обчисліть:

````python
# Код студента
import numpy as np

A = np.array([
    [4, 7],
    [2, 6]
])

det = np.linalg.det(A)

A_inv = np.linalg.inv(A)

check = A @ A_inv

print("Визначник:", det)

print("\nОбернена матриця:")
print(A_inv)

print("\nПеревірка A @ A_inv:")
print(check)
````

### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
import numpy as np

A = np.array([
    [10, 20, 30],
    [20, 30, 40],
    [30, 40, 50]
])

mu = np.mean(A, axis=0)
sigma = np.std(A, axis=0)

Z = (A - mu) / sigma

print(Z)
````

Дано результати продажів трьох товарів за чотири місяці:

```text
sales = [[120, 150, 170, 160],
         [80,  90,  100, 110],
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
# Код студента
import numpy as np

sales = np.array([
    [120, 150, 170, 160],
    [80,  90,  100, 110],
    [200, 180, 210, 230]
])

total_by_product = np.sum(sales, axis=1)

total_by_month = np.sum(sales, axis=0)

best_product = np.argmax(total_by_product) + 1

best_month = np.argmax(total_by_month) + 1

average_by_product = np.mean(sales, axis=1)

print("Продажі кожного товару:", total_by_product)
print("Продажі кожного місяця:", total_by_month)
print("Товар з найбільшими продажами:", best_product)
print("Місяць з найбільшими продажами:", best_month)
print("Середні продажі кожного товару:", average_by_product)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
import numpy as np

np.random.seed(42)

A = np.random.randint(1, 21, (5, 5))

print("Матриця:")
print(A)

main_diag_sum = np.trace(A)

secondary_diag_sum = np.trace(np.fliplr(A))

row_max = np.max(A, axis=1)

col_min = np.min(A, axis=0)

mean_value = np.mean(A)
A[A < mean_value] = 0

print("Сума головної діагоналі:", main_diag_sum)
print("Сума побічної діагоналі:", secondary_diag_sum)
print("Максимуми рядків:", row_max)
print("Мінімуми стовпців:", col_min)
print("Середнє значення:", mean_value)
print("Матриця після заміни:")
print(A)
````

```text
X = [[2, 4, 6],
```

Обчисліть **евклідову відстань** між:

- другим і третім рядком.

### Завдання 39
Створіть матрицю `4 × 4` із довільних цілих чисел.

Для кожного рядка знайдіть його норму та створіть нову матрицю, у якій кожен рядок нормалізований до одиничної довжини.

Спробуйте виконати завдання без циклів.

````python
# Код студента
import numpy as np

np.random.seed(42)

A = np.random.randint(1, 21, (4, 4))

norms = np.linalg.norm(A, axis=1)

normalized_A = A / norms[:, np.newaxis]

print("Матриця:")
print(A)

print("\nНорми рядків:")
print(norms)

print("\nНормалізована матриця:")
print(normalized_A)

print("\nНорми нормалізованих рядків:")
print(np.linalg.norm(normalized_A, axis=1))
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
import numpy as np

A = np.array([
    [1, 2, 1],
    [2, 1, -1],
    [3, -1, 2]
])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)

result = A @ x

comparison = np.allclose(result, b)

print("Розв'язок:", x)
print("A @ x:", result)
print("Вектор b:", b)
print("Результати збігаються:", comparison)
````
